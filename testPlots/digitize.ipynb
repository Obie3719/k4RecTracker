```python
# Quick event display helpers for simulated vs digitized DCH hits
import itertools

# choose the events you would like to inspect closely
example_events = [27, 35, 53, 83, 86, 99]

# guard against requesting events that are not present
n_events = len(dcPos_x)
valid_events = [evt for evt in example_events if 0 <= evt < n_events]
if len(valid_events) != len(example_events):
    missing = set(example_events) - set(valid_events)
    print(f"Skipping out-of-range events: {sorted(missing)}")

def plot_dch_event(event_idx, output_dir='.', show_plot=True):
    """Create simple XY and ZR projections for one event."""
    sim_x = ak.to_numpy(dcPos_x[event_idx])
    sim_y = ak.to_numpy(dcPos_y[event_idx])
    sim_z = ak.to_numpy(dcPos_z[event_idx])

    digi_x = ak.to_numpy(digiPos_x[event_idx])
    digi_y = ak.to_numpy(digiPos_y[event_idx])
    digi_z = ak.to_numpy(digiPos_z[event_idx])

    sim_r = np.sqrt(sim_x**2 + sim_y**2)
    digi_r = np.sqrt(digi_x**2 + digi_y**2)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    # XY view (transverse plane)
    axes[0].scatter(sim_x, sim_y, s=25, c='tab:blue', label='Sim hit', alpha=0.7)
    axes[0].scatter(digi_x, digi_y, s=25, c='tab:orange', marker='x', label='Digi hit')
    for sx, sy, dx, dy in zip(sim_x, sim_y, digi_x, digi_y):
        axes[0].plot([sx, dx], [sy, dy], color='gray', alpha=0.2)
    axes[0].set_xlabel('x [mm]')
    axes[0].set_ylabel('y [mm]')
    axes[0].set_title(f'DCH transverse view – event {event_idx}')
    axes[0].axhline(0, color='black', lw=0.5, alpha=0.3)
    axes[0].axvline(0, color='black', lw=0.5, alpha=0.3)
    axes[0].legend(loc='upper right')
    axes[0].set_aspect('equal', adjustable='box')

    # ZR view (longitudinal)
    axes[1].scatter(sim_z, sim_r, s=25, c='tab:blue', label='Sim hit', alpha=0.7)
    axes[1].scatter(digi_z, digi_r, s=25, c='tab:orange', marker='x', label='Digi hit')
    for sz, sr, dz, dr in zip(sim_z, sim_r, digi_z, digi_r):
        axes[1].plot([sz, dz], [sr, dr], color='gray', alpha=0.2)
    axes[1].set_xlabel('z [mm]')
    axes[1].set_ylabel('r [mm]')
    axes[1].set_title(f'DCH longitudinal view – event {event_idx}')
    axes[1].legend(loc='upper right')

    fig.tight_layout()
    output_path = f"{output_dir}/event_display_DCH_evt{event_idx:04d}.png"
    fig.savefig(output_path, dpi=200)
    if show_plot:
        plt.show()
    else:
        plt.close(fig)
    return output_path

saved_displays = []
for evt in valid_events:
    saved_displays.append(plot_dch_event(evt, output_dir='.', show_plot=True))

print('Saved displays:', saved_displays)
```
